```python
#import librarys
import pandas as pd
import sqlalchemy as sq

from sqlalchemy import text
from sqlalchemy import create_engine
```

```python
#import new york csv
nyc_data_2025 = pd.read_csv("Data/nyc_2025_clean_w_response_time.csv", 
                            dtype = {"finalcalltypedescription" : "category"})



# take a random sample of new york
nyc_data_2025 = nyc_data_2025.sample(n = 200000, random_state = 42)
```

```python
#import seatle
seattle_data = pd.read_csv("Data/seattle_2025_cleaned.csv", 
                           dtype = {"finalcalltypedescription" : "category"})
#import waterloo
waterloo_data = pd.read_csv("Data/waterloo_clean.csv", 
                           dtype = {"finalcalltypedescription" : "category"})
#import san-francisco
san_francisco = pd.read_csv("Data/san_francisco_cleaned.csv", 
                           dtype = {"finalcalltypedescription" : "category"})
```

```python
#open database password
with open("dbpasswd", "r", encoding = "utf-8") as f:
    
    PWD = f.read().strip() 

USER = "student"
DB   = "student"

#make database engine
sql_engine = create_engine(
    f"mysql+mysqlconnector://{USER}:{PWD}@127.0.0.1:3306/{DB}")
```

```python
#read waterloo data to engine
waterloo_data.to_sql("waterloo_emergency_call_data_2025", 
                     sql_engine, 
                     index = False, 
                     if_exists = 'replace', 
                     chunksize = 1000)
```

```python
#read seatle data to engine
seattle_data.to_sql("seattle_emergency_call_data_2025", 
                     sql_engine, 
                     index = False, 
                     if_exists = 'replace', 
                     chunksize = 1000)
```

```python
#read san-fransico data to engine
san_francisco.to_sql("san_francisco_emergency_call_data_2025", 
                     sql_engine, 
                     index = False, 
                     if_exists = 'replace', 
                     chunksize = 1000)
```

```python
#read new york data to engine
nyc_data_2025.to_sql("new_york_emergency_call_data_2025", 
                     sql_engine, 
                     index = False, 
                     if_exists = 'replace', 
                     chunksize = 1000)
```

```python
#perfrom a union on all the tables
query_union = '''
SELECT * FROM seattle_emergency_call_data_2025
UNION ALL
SELECT * FROM san_francisco_emergency_call_data_2025
UNION ALL
SELECT * FROM waterloo_emergency_call_data_2025
UNION ALL
SELECT * FROM new_york_emergency_call_data_2025;
AS combined_response_time
'''

query_union_result = pd.read_sql(text(query_union), sql_engine)

query_union_result

query_union_result.shape

sql_engine.dispose()
```
